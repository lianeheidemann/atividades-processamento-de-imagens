# 1. Importar as bibliotecas

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image
from google.colab import files
```

# 2. Carregar a imagem no Colab

```python
arquivos = files.upload()

nome_arquivo = next(iter(arquivos))
imagem = Image.open(nome_arquivo)

imagem
```

# 3. Converter a imagem para níveis de cinza

```python
imagem_cinza = imagem.convert("L")

plt.figure(figsize=(7, 5))
plt.imshow(imagem_cinza, cmap="gray")
plt.title("Imagem em níveis de cinza")
plt.axis("off")
plt.show()
```

# 4. Converter a imagem em uma matriz

```python
matriz = np.array(imagem_cinza)

print("Dimensões da matriz:", matriz.shape)
print(matriz)
```

Cada posição da matriz contém uma intensidade entre:

* 0: preto;
* 255: branco;
* valores intermediários: tons de cinza.

# 5. Criar o vetor e contar os pixels

```python
# Uma posição para cada intensidade: 0 até 255
histograma = np.zeros(256, dtype=int)

# Percorrer cada pixel da imagem
for linha in matriz:
    for pixel in linha:
        intensidade = int(pixel)
        histograma[intensidade] += 1

print(histograma)
```

# 6. Mostrar o histograma com barplot()

```python
intensidades = np.arange(256)

plt.figure(figsize=(16, 6))

sns.barplot(
    x=intensidades,
    y=histograma,
    color="steelblue"
)

plt.title("Histograma da imagem em níveis de cinza")
plt.xlabel("Intensidade")
plt.ylabel("Quantidade de pixels")

# Mostrar somente alguns números no eixo X para não ficar poluído
plt.xticks(
    ticks=np.arange(0, 256, 16),
    labels=np.arange(0, 256, 16)
)

plt.show()
```
