# 1. Importar as bibliotecas e carregar imagens

```python
# Importar as bibliotecas
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
from io import BytesIO

from PIL import Image
from IPython.display import display
#from google.colab import files
```

```python
# Endereço da pasta com as imagens no GitHub
url_base = (
    "https://raw.githubusercontent.com/"
    "lianeheidemann/atividades-processamento-de-imagens/"
    "main/atividade/atividade-2/input/"
)

# Função para carregar uma imagem diretamente do GitHub
def carregar_imagem(nome_arquivo):
    resposta = requests.get(url_base + nome_arquivo)
    resposta.raise_for_status()
    return Image.open(BytesIO(resposta.content))

# Carregar as três imagens
imagem_1 = carregar_imagem("input_1.png")
imagem_2 = carregar_imagem("input_2.png")
imagem_3 = carregar_imagem("input_3.png")

# Criar uma pasta para salvar os resultados no Colab
pasta_saida = "/content/output"
os.makedirs(pasta_saida, exist_ok=True)

print("Imagens carregadas corretamente.")
```

```python
# Exibir as três imagens
fig = plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(imagem_1)
plt.title("Mesa à noite")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(imagem_2)
plt.title("Orla urbana")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(imagem_3)
plt.title("Dália")
plt.axis("off")

plt.tight_layout()
fig.savefig(os.path.join(pasta_saida, "imagens_originais.png"))
plt.show()
```

# 2. Converter para níveis de cinza

```python
# Converter as três imagens para tons de cinza
imagem_cinza_1 = imagem_1.convert("L")
imagem_cinza_2 = imagem_2.convert("L")
imagem_cinza_3 = imagem_3.convert("L")
```

```python
# Exibir as três imagens
plt.figure(figsize=(15, 5))

# Criar a área de exibição
fig = plt.figure(figsize=(15, 5))

# Primeira imagem
plt.subplot(1, 3, 1)
plt.imshow(imagem_cinza_1, cmap="gray", vmin=0, vmax=255)
plt.title("Imagem 1 — escura")
plt.axis("off")

# Segunda imagem
plt.subplot(1, 3, 2)
plt.imshow(imagem_cinza_2, cmap="gray", vmin=0, vmax=255)
plt.title("Imagem 2 — clara")
plt.axis("off")

# Terceira imagem
plt.subplot(1, 3, 3)
plt.imshow(imagem_cinza_3, cmap="gray", vmin=0, vmax=255)
plt.title("Imagem 3 — tons médios")
plt.axis("off")

plt.tight_layout()
fig.savefig(os.path.join(pasta_saida, "imagens_cinza.png"))
plt.show()
```

# 3. Histograma

```python
fig = plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
plt.hist(
    np.array(imagem_cinza_1).ravel(),
    bins=256,
    range=(0, 256),
    color="gray"
)
plt.title("Histograma — imagem escura")
plt.xlabel("Intensidade")
plt.ylabel("Quantidade de pixels")
plt.xlim(0, 255)

plt.subplot(1, 3, 2)
plt.hist(
    np.array(imagem_cinza_2).ravel(),
    bins=256,
    range=(0, 256),
    color="gray"
)
plt.title("Histograma — imagem clara")
plt.xlabel("Intensidade")
plt.ylabel("Quantidade de pixels")
plt.xlim(0, 255)

plt.subplot(1, 3, 3)
plt.hist(
    np.array(imagem_cinza_3).ravel(),
    bins=256,
    range=(0, 256),
    color="gray"
)
plt.title("Histograma — imagem")
plt.xlabel("Intensidade")
plt.ylabel("Quantidade de pixels")
plt.xlim(0, 255)

plt.tight_layout()
fig.savefig(os.path.join(pasta_saida, "histogramas.png"))
plt.show()
```

- Imagem 1: As intensidades estão fortemente concentradas entre 0 e 60 (93,5% dos pixels), com média 13,7, indicando uma imagem muito escura, sem uso da faixa superior de intensidades.
- Imagem 2: Há uma forte concentração acima de 200 (81,5% dos pixels), com média 229,3, indicando predominância de regiões muito claras, principalmente por causa do céu estourado ao fundo.
- Imagem 3: As intensidades estão amplamente distribuídas entre 7 e 255, com média 113,3, indicando uma imagem de tons médios com boa variedade de intensidades.

# 5. Correção de gamma

```python
# Converter as imagens para matrizes NumPy
matriz_1 = np.array(imagem_cinza_1, dtype=np.uint8)
matriz_2 = np.array(imagem_cinza_2, dtype=np.uint8)
matriz_3 = np.array(imagem_cinza_3, dtype=np.uint8)

# Parâmetro da correção de gamma
gamma = 0.5

# Aplicar a mesma operação nas três imagens
resultado_1 = (255 * ((matriz_1 / 255.0) ** gamma)).astype(np.uint8)
resultado_2 = (255 * ((matriz_2 / 255.0) ** gamma)).astype(np.uint8)
resultado_3 = (255 * ((matriz_3 / 255.0) ** gamma)).astype(np.uint8)
```

```python
# Organizar as imagens para facilitar a construção da figura
originais = [matriz_1, matriz_2, matriz_3]
resultados = [resultado_1, resultado_2, resultado_3]

nomes = [
    "Imagem 1 — escura",
    "Imagem 2 — clara",
    "Imagem 3 — tons médios"
]

# Criar uma figura com 3 linhas e 4 colunas
fig, eixos = plt.subplots(3, 4, figsize=(20, 15))

for i in range(3):

    # Imagem antes da operação
    eixos[i, 0].imshow(
        originais[i],
        cmap="gray",
        vmin=0,
        vmax=255
    )
    eixos[i, 0].set_title(f"{nomes[i]}\nAntes")
    eixos[i, 0].axis("off")

    # Imagem depois da operação
    eixos[i, 1].imshow(
        resultados[i],
        cmap="gray",
        vmin=0,
        vmax=255
    )
    eixos[i, 1].set_title(f"{nomes[i]}\nDepois (γ = {gamma})")
    eixos[i, 1].axis("off")

    # Histograma antes da operação
    eixos[i, 2].hist(
        originais[i].ravel(),
        bins=256,
        range=(0, 256),
        color="gray"
    )
    eixos[i, 2].set_title("Histograma antes")
    eixos[i, 2].set_xlabel("Intensidade")
    eixos[i, 2].set_ylabel("Pixels")
    eixos[i, 2].set_xlim(0, 255)

    # Histograma depois da operação
    eixos[i, 3].hist(
        resultados[i].ravel(),
        bins=256,
        range=(0, 256),
        color="steelblue"
    )
    eixos[i, 3].set_title("Histograma depois")
    eixos[i, 3].set_xlabel("Intensidade")
    eixos[i, 3].set_ylabel("Pixels")
    eixos[i, 3].set_xlim(0, 255)

plt.suptitle(
    "Correção de gamma nas três imagens (γ = 0,5)",
    fontsize=18
)

plt.tight_layout(rect=[0, 0, 1, 0.97])
fig.savefig(os.path.join(pasta_saida, "correcao_gamma.png"))
plt.show()
```

- A correção de gamma fez menos diferença na Imagem 2, com uma diferença média de intensidade de apenas 11,2 níveis, contra 32,5 na Imagem 1 e 51,0 na Imagem 3. Isso acontece porque seu histograma original já estava fortemente concentrado nas intensidades altas (81,5% dos pixels acima de 200), indicando que ela já era predominantemente clara. Como foi utilizado γ = 0,5, os pixels foram deslocados em direção ao branco, mas as intensidades dessa imagem já estavam próximas do valor máximo de 255; por isso, a alteração visual foi menor do que nas imagens 1 e 3.
