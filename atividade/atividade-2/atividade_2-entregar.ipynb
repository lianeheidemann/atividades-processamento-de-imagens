# 1. Importar as bibliotecas e carregar imagens

```python
# Importar as bibliotecas
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from PIL import Image
from IPython.display import display
#from google.colab import files
```

```python
# Caminhos das imagens no computador
# Clara / Escura / Bem
nome_arquivo_1 = r"C:\Users\Lily\Documents\GitHub\atividades-processamento-de-imagens\atividade\atividade-2\input\input_1.png"
nome_arquivo_2 = r"C:\Users\Lily\Documents\GitHub\atividades-processamento-de-imagens\atividade\atividade-2\input\input_2.png"
nome_arquivo_3 = r"C:\Users\Lily\Documents\GitHub\atividades-processamento-de-imagens\atividade\atividade-2\input\input_3.png"

# Abrir as imagens
imagem_1 = Image.open(nome_arquivo_1)
imagem_2 = Image.open(nome_arquivo_2)
imagem_3 = Image.open(nome_arquivo_3)

# Pasta de saída das imagens geradas
pasta_saida = os.path.join(os.path.dirname(os.path.dirname(nome_arquivo_1)), "output")
os.makedirs(pasta_saida, exist_ok=True)
```

```python
# Exibir as três imagens
fig = plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(imagem_1)
plt.title("Mandala")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(imagem_2)
plt.title("Sorvete")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(imagem_3)
plt.title("Cute")
plt.axis("off")

plt.tight_layout()
fig.savefig(os.path.join(pasta_saida, "imagens_originais.png"))
plt.show()
```

# 2. Converter para níveis de cinza

```python
# Converter as três imagens para tons de cinza
imagem_cinza_1 = imagem_1.convert("L")
imagem_cinza_2 = imagem_2.convert("L")
imagem_cinza_3 = imagem_3.convert("L")
```

```python
# Exibir as três imagens
plt.figure(figsize=(15, 5))

# Criar a área de exibição
fig = plt.figure(figsize=(15, 5))

# Primeira imagem
plt.subplot(1, 3, 1)
plt.imshow(imagem_cinza_1, cmap="gray", vmin=0, vmax=255)
plt.title("Imagem 1 — clara")
plt.axis("off")

# Segunda imagem
plt.subplot(1, 3, 2)
plt.imshow(imagem_cinza_2, cmap="gray", vmin=0, vmax=255)
plt.title("Imagem 2 — escura")
plt.axis("off")

# Terceira imagem
plt.subplot(1, 3, 3)
plt.imshow(imagem_cinza_3, cmap="gray", vmin=0, vmax=255)
plt.title("Imagem 3 — bem distribuída")
plt.axis("off")

plt.tight_layout()
fig.savefig(os.path.join(pasta_saida, "imagens_cinza.png"))
plt.show()
```

# 3. Histograma

```python
fig = plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
plt.hist(
    np.array(imagem_cinza_1).ravel(),
    bins=256,
    range=(0, 256),
    color="gray"
)
plt.title("Histograma — imagem escura")
plt.xlabel("Intensidade")
plt.ylabel("Quantidade de pixels")
plt.xlim(0, 255)

plt.subplot(1, 3, 2)
plt.hist(
    np.array(imagem_cinza_2).ravel(),
    bins=256,
    range=(0, 256),
    color="gray"
)
plt.title("Histograma — imagem clara")
plt.xlabel("Intensidade")
plt.ylabel("Quantidade de pixels")
plt.xlim(0, 255)

plt.subplot(1, 3, 3)
plt.hist(
    np.array(imagem_cinza_3).ravel(),
    bins=256,
    range=(0, 256),
    color="gray"
)
plt.title("Histograma — imagem")
plt.xlabel("Intensidade")
plt.ylabel("Quantidade de pixels")
plt.xlim(0, 255)

plt.tight_layout()
fig.savefig(os.path.join(pasta_saida, "histogramas.png"))
plt.show()
```

- Imagem 1: As intensidades estão concentradas principalmente entre aproximadamente 10 e 60, indicando uma imagem predominantemente escura, embora existam alguns pixels claros.
- Imagem 2: Há uma forte concentração próxima de 230, indicando predominância de regiões muito claras, principalmente por causa do fundo branco e quadriculado.
- Imagem 3: As intensidades concentram-se principalmente entre 150 e 180, indicando uma imagem relativamente clara, com pouco contraste e predominância de tons médios-claros.

# 5. Correção de gamma

```python
# Converter as imagens para matrizes NumPy
matriz_1 = np.array(imagem_cinza_1, dtype=np.uint8)
matriz_2 = np.array(imagem_cinza_2, dtype=np.uint8)
matriz_3 = np.array(imagem_cinza_3, dtype=np.uint8)

# Parâmetro da correção de gamma
gamma = 0.5

# Aplicar a mesma operação nas três imagens
resultado_1 = (255 * ((matriz_1 / 255.0) ** gamma)).astype(np.uint8)
resultado_2 = (255 * ((matriz_2 / 255.0) ** gamma)).astype(np.uint8)
resultado_3 = (255 * ((matriz_3 / 255.0) ** gamma)).astype(np.uint8)
```

```python
# Organizar as imagens para facilitar a construção da figura
originais = [matriz_1, matriz_2, matriz_3]
resultados = [resultado_1, resultado_2, resultado_3]

nomes = [
    "Imagem 1 — escura",
    "Imagem 2 — clara",
    "Imagem 3 — tons médios"
]

# Criar uma figura com 3 linhas e 4 colunas
fig, eixos = plt.subplots(3, 4, figsize=(20, 15))

for i in range(3):

    # Imagem antes da operação
    eixos[i, 0].imshow(
        originais[i],
        cmap="gray",
        vmin=0,
        vmax=255
    )
    eixos[i, 0].set_title(f"{nomes[i]}\nAntes")
    eixos[i, 0].axis("off")

    # Imagem depois da operação
    eixos[i, 1].imshow(
        resultados[i],
        cmap="gray",
        vmin=0,
        vmax=255
    )
    eixos[i, 1].set_title(f"{nomes[i]}\nDepois (γ = {gamma})")
    eixos[i, 1].axis("off")

    # Histograma antes da operação
    eixos[i, 2].hist(
        originais[i].ravel(),
        bins=256,
        range=(0, 256),
        color="gray"
    )
    eixos[i, 2].set_title("Histograma antes")
    eixos[i, 2].set_xlabel("Intensidade")
    eixos[i, 2].set_ylabel("Pixels")
    eixos[i, 2].set_xlim(0, 255)

    # Histograma depois da operação
    eixos[i, 3].hist(
        resultados[i].ravel(),
        bins=256,
        range=(0, 256),
        color="steelblue"
    )
    eixos[i, 3].set_title("Histograma depois")
    eixos[i, 3].set_xlabel("Intensidade")
    eixos[i, 3].set_ylabel("Pixels")
    eixos[i, 3].set_xlim(0, 255)

plt.suptitle(
    "Correção de gamma nas três imagens (γ = 0,5)",
    fontsize=18
)

plt.tight_layout(rect=[0, 0, 1, 0.97])
fig.savefig(os.path.join(pasta_saida, "correcao_gamma.png"))
plt.show()
```

- A correção de gamma fez menos diferença na Imagem 2, pois seu histograma original já estava fortemente concentrado nas intensidades altas, principalmente próximo de 230, indicando que ela já era predominantemente clara. Como foi utilizado γ = 0,5, os pixels foram deslocados em direção ao branco, mas as intensidades dessa imagem já estavam próximas do valor máximo de 255; por isso, a alteração visual foi menor do que nas imagens 1 e 3.
